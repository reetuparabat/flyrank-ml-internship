## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

```python
## My lane as an ML task

'''My lane is best framed as a **classification** problem.
Each content page can be classified based on whether it needs a content refresh.
The goal is to identify pages that are likely to require updates using information already available in the dataset, instead of reviewing every page manually.
The output supports prioritizing which pages should be refreshed first.'''
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
## 2. Target or Proxy

'''The target is to predict whether a content page is declining (`is_declining_label`).
This label is not directly provided by a human. Instead, it is created using a defined rule based on the `trend_direction` column.
If `trend_direction` is "down", the page is labeled as declining; otherwise, it is labeled as not declining.
This is a proxy target because it represents the page's current performance rather than predicting future decline.
In a real-world system, a stronger target would be whether a page is likely to decline in the future, allowing editors to refresh content before its performance drops.'''
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
## 3. Success Metric

'''I will use **F1 Score** as the primary success metric because the task is to identify declining content pages while balancing false positives and false negatives.
A model with high precision but low recall would miss pages that need refreshing, while a model with high recall but low precision would recommend refreshing too many unnecessary pages. F1 Score provides a balanced measure of both.
For an initial model, an F1 Score above 0.80 would indicate useful performance, although the final evaluation should always be compared against a baseline model.'''
```

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
## Unit of Analysis

'''Each row in the dataframe represents one content page.
The row contains features describing that page, such as impressions, CTR, engagement, content age, days since the last update, and trend direction.
The model uses these features to predict whether that specific content page is declining and may need to be refreshed.'''
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
'''A fixed rule is difficult because content performance depends on multiple interacting factors such as CTR, impressions, engagement rate, content age, and days since the last update.
A simple if-statement cannot capture all these relationships or handle the different combinations that occur across thousands of content pages.
A machine learning model can learn these complex patterns from historical data and make better predictions than a manually written set of rules.'''
```
